```python
from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import mm
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
    PageBreak, KeepTogether
)
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics
from reportlab.lib.colors import HexColor
import os
os.makedirs("/mnt/data", exist_ok=True)

out = "/mnt/data/IMLC_2026_Qualification_Solutions.pdf"

styles = getSampleStyleSheet()
styles.add(ParagraphStyle(
    name="DocTitle", parent=styles["Title"], fontName="Helvetica-Bold",
    fontSize=19, leading=23, alignment=TA_CENTER, textColor=HexColor("#17324D"),
    spaceAfter=5
))
styles.add(ParagraphStyle(
    name="SubTitle", parent=styles["Normal"], fontName="Helvetica",
    fontSize=10, leading=14, alignment=TA_CENTER, textColor=HexColor("#526579"),
    spaceAfter=14
))
styles.add(ParagraphStyle(
    name="Problem", parent=styles["Heading1"], fontName="Helvetica-Bold",
    fontSize=14, leading=18, textColor=HexColor("#17324D"),
    spaceBefore=10, spaceAfter=7
))
styles.add(ParagraphStyle(
    name="Part", parent=styles["Heading2"], fontName="Helvetica-Bold",
    fontSize=10.5, leading=14, textColor=HexColor("#245B82"),
    spaceBefore=6, spaceAfter=4
))
styles.add(ParagraphStyle(
    name="Body2", parent=styles["BodyText"], fontName="Helvetica",
    fontSize=9.4, leading=13.4, spaceAfter=5
))
styles.add(ParagraphStyle(
    name="Small", parent=styles["BodyText"], fontName="Helvetica",
    fontSize=8.5, leading=11, spaceAfter=3
))
styles.add(ParagraphStyle(
    name="Answer", parent=styles["BodyText"], fontName="Helvetica-Bold",
    fontSize=9.5, leading=13, textColor=HexColor("#17324D"),
    backColor=HexColor("#EEF4F8"), borderPadding=6, spaceBefore=3, spaceAfter=7
))
styles.add(ParagraphStyle(
    name="Equation", parent=styles["BodyText"], fontName="Courier",
    fontSize=9, leading=13, leftIndent=12, spaceBefore=2, spaceAfter=4
))

story = []
P = lambda txt, style="Body2": Paragraph(txt, styles[style])
story += [
    Spacer(1, 8*mm),
    P("INTERNATIONAL MACHINE LEARNING COMPETITION", "DocTitle"),
    P("Edition 2026 · Qualification Round · Written Solutions", "SubTitle"),
]
identity = Table([
    [P("<b>Name:</b> ____________________________________", "Body2"),
     P("<b>Age group:</b> ____________________", "Body2")],
], colWidths=[100*mm, 75*mm])
identity.setStyle(TableStyle([
    ("VALIGN",(0,0),(-1,-1),"MIDDLE"),
    ("BOTTOMPADDING",(0,0),(-1,-1),8),
]))
story += [identity, Spacer(1, 3*mm)]

story += [
    P("Problem A – Machine Learning Lifecycle (5 points)", "Problem"),
    P("<b>(a) Labels for each step</b>", "Part"),
]
data = [
    [P("<b>Step</b>","Small"), P("<b>Description</b>","Small"), P("<b>Label</b>","Small")],
    [P("1","Small"), P("Collect 10,000 recordings labelled speech, music, or alarm.","Small"), P("Data collection","Small")],
    [P("2","Small"), P("Adjust the model’s parameters using these recordings.","Small"), P("Training","Small")],
    [P("3","Small"), P("Test the model on recordings it has not seen before.","Small"), P("Evaluation","Small")],
    [P("4","Small"), P("Install the model in a school.","Small"), P("Deployment","Small")],
    [P("5","Small"), P("The installed model receives a new sound and predicts “alarm”.","Small"), P("Inference","Small")],
    [P("6","Small"), P("Label newly recorded sounds and use them to adjust the model again.","Small"), P("Data collection and training","Small")],
]
t = Table(data, colWidths=[16*mm, 105*mm, 54*mm], repeatRows=1)
t.setStyle(TableStyle([
    ("BACKGROUND",(0,0),(-1,0),HexColor("#E5EEF5")),
    ("GRID",(0,0),(-1,-1),0.4,HexColor("#B7C7D3")),
    ("VALIGN",(0,0),(-1,-1),"TOP"),
    ("LEFTPADDING",(0,0),(-1,-1),6),("RIGHTPADDING",(0,0),(-1,-1),6),
    ("TOPPADDING",(0,0),(-1,-1),5),("BOTTOMPADDING",(0,0),(-1,-1),5),
]))
story += [t, Spacer(1, 3*mm), P("<b>(b) When does the system learn?</b>", "Part"),
          P("The system actually learns during Steps 2 and 6. Learning means changing the model’s internal parameters based on labelled examples so that its predictions improve. Step 2 uses the initial recordings for training. In Step 6, newly collected and labelled recordings are used for further training, allowing the model to adapt to new data.", "Body2")]

story += [
    P("Problem B – The Greenhouse Tree (5 points)", "Problem"),
    P("<b>(a) Temperature = 26 °C and humidity = 68%</b>", "Part"),
    P("The temperature is not above 28 °C, so follow the “No” branch. The humidity is also not above 70%, so follow its “No” branch.", "Body2"),
    P("Predicted action: KEEP CLOSED.", "Answer"),
    P("<b>(b) Adjusted decision tree</b>", "Part"),
    P("One decision tree that correctly classifies all six observations is:", "Body2"),
]
tree_data = [
    [P("<b>Condition</b>","Small"), P("<b>Branch</b>","Small"), P("<b>Next step / prediction</b>","Small")],
    [P("Temperature > 28 °C?","Small"), P("Yes","Small"), P("<b>OPEN</b>","Small")],
    [P("Temperature > 28 °C?","Small"), P("No","Small"), P("Check whether humidity > 70%","Small")],
    [P("Humidity > 70%?","Small"), P("Yes","Small"), P("<b>OPEN</b>","Small")],
    [P("Humidity > 70%?","Small"), P("No","Small"), P("Check whether CO2 > 1,200 ppm?","Small")],
    [P("CO2 > 1,200 ppm?","Small"), P("Yes","Small"), P("<b>OPEN</b>","Small")],
    [P("CO2 > 1,200 ppm?","Small"), P("No","Small"), P("<b>KEEP CLOSED</b>","Small")],
]
tt = Table(tree_data, colWidths=[54*mm, 25*mm, 96*mm], repeatRows=1)
tt.setStyle(TableStyle([
    ("BACKGROUND",(0,0),(-1,0),HexColor("#E5EEF5")),
    ("GRID",(0,0),(-1,-1),0.4,HexColor("#B7C7D3")),
    ("VALIGN",(0,0),(-1,-1),"TOP"),
    ("LEFTPADDING",(0,0),(-1,-1),6),("RIGHTPADDING",(0,0),(-1,-1),6),
    ("TOPPADDING",(0,0),(-1,-1),5),("BOTTOMPADDING",(0,0),(-1,-1),5),
]))
story += [tt, Spacer(1, 2*mm), P("<b>Verification against the observations</b>", "Part")]
verify = [
    [P("<b>Temperature</b>","Small"), P("<b>Humidity</b>","Small"), P("<b>CO2</b>","Small"), P("<b>Prediction</b>","Small"), P("<b>Logged action</b>","Small")],
    [P("31 °C","Small"),P("60%","Small"),P("700 ppm","Small"),P("OPEN","Small"),P("OPEN","Small")],
    [P("26 °C","Small"),P("75%","Small"),P("650 ppm","Small"),P("OPEN","Small"),P("OPEN","Small")],
    [P("25 °C","Small"),P("60%","Small"),P("800 ppm","Small"),P("KEEP CLOSED","Small"),P("KEEP CLOSED","Small")],
    [P("27 °C","Small"),P("68%","Small"),P("1,100 ppm","Small"),P("KEEP CLOSED","Small"),P("KEEP CLOSED","Small")],
    [P("24 °C","Small"),P("55%","Small"),P("1,400 ppm","Small"),P("OPEN","Small"),P("OPEN","Small")],
    [P("22 °C","Small"),P("50%","Small"),P("1,550 ppm","Small"),P("OPEN","Small"),P("OPEN","Small")],
]
vt = Table(verify, colWidths=[27*mm,23*mm,31*mm,47*mm,47*mm], repeatRows=1)
vt.setStyle(TableStyle([
    ("BACKGROUND",(0,0),(-1,0),HexColor("#E5EEF5")),
    ("GRID",(0,0),(-1,-1),0.4,HexColor("#B7C7D3")),
    ("VALIGN",(0,0),(-1,-1),"MIDDLE"),
    ("LEFTPADDING",(0,0),(-1,-1),4),("RIGHTPADDING",(0,0),(-1,-1),4),
    ("TOPPADDING",(0,0),(-1,-1),4),("BOTTOMPADDING",(0,0),(-1,-1),4),
]))
story += [vt, P("All six predictions match the actions in the log.", "Body2")]

story += [
    P("Problem C – To Fit or Not to Fit (5 points)", "Problem"),
    P("The score is J = Σ(y − ŷ)² + λΣaᵢ², with λ = 1. The regularisation sum includes the coefficients of x, x² and x³; a missing coefficient is zero.", "Body2"),
    P("<b>(a) Calculate the score for each model</b>", "Part"),
    P("<b>Model M1:</b> M1(x) = 0.2x³ − 0.9x² + 2.9x + 1", "Body2"),
    P("For x = 0, 1, 2, 3, the predictions are 1, 3.2, 4.8, 7. These equal the observed y-values, so the sum of squared errors is 0.", "Body2"),
    P("Regularisation penalty = (2.9)² + (−0.9)² + (0.2)² = 8.41 + 0.81 + 0.04 = 9.26.", "Equation"),
    P("J(M1) = 0 + 1(9.26) = 9.26.", "Answer"),
    P("<b>Model M2:</b> M2(x) = 2x + 1", "Body2"),
    P("For x = 0, 1, 2, 3, the predictions are 1, 3, 5, 7. The sum of squared errors is (1−1)² + (3.2−3)² + (4.8−5)² + (7−7)² = 0 + 0.04 + 0.04 + 0 = 0.08.", "Body2"),
    P("The only non-zero coefficient among x, x² and x³ is 2. The regularisation penalty is 2² = 4.", "Body2"),
    P("J(M2) = 0.08 + 1(4) = 4.08.", "Answer"),
    P("The scoring rule selects M2 because 4.08 < 9.26.", "Body2"),
    P("<b>(b) Fit and purpose of regularisation</b>", "Part"),
    P("M1 fits the training data better because it predicts all four training values exactly, giving zero squared error. Nevertheless, the scoring rule selects M2 because M1 has larger coefficients and therefore receives a larger regularisation penalty. The term λΣaᵢ² is a regularisation penalty: it discourages large coefficients and can reduce overfitting, encouraging a model that generalises better to unseen data.", "Body2"),
]

story += [
    P("Problem D – The Price of Drift (5 points)", "Problem"),
    P("The loss is L(t) = −rt + βt², where t ≥ 0 and r, β > 0.", "Body2"),
    P("<b>(a) Find the minimizing shift and minimum loss</b>", "Part"),
    P("Differentiate: dL/dt = −r + 2βt. Set the derivative equal to zero: −r + 2βt = 0, giving t* = r/(2β). Since d²L/dt² = 2β > 0, this stationary point is a minimum.", "Body2"),
    P("Substitute t* into the loss: L(t*) = −r(r/(2β)) + β(r/(2β))² = −r²/(2β) + r²/(4β) = −r²/(4β).", "Equation"),
    P("Therefore, t* = r/(2β) and L(t*) = −r²/(4β).", "Answer"),
    P("<b>(b) Limits as β approaches zero and infinity</b>", "Part"),
    P("As β → 0, t* = r/(2β) → ∞. The drift penalty becomes negligible, so the model can move very far from its reference model to increase reward. As β → ∞, t* → 0. A very large penalty keeps the model close to its trusted reference model.", "Body2"),
    P("<b>(c) Prove the safety condition</b>", "Part"),
    P("The required condition is t* ≤ T for every reward satisfying 0 < r ≤ rmax. Since t* = r/(2β), this is equivalent to r/(2β) ≤ T, or r ≤ 2βT. For this to hold for every allowed r, it must hold at the largest possible reward, r = rmax. Thus rmax ≤ 2βT, which is equivalent to β ≥ rmax/(2T).", "Body2"),
    P("Therefore, the shift is guaranteed to remain at or below T for every permitted reward exactly when β ≥ rmax/(2T).", "Answer"),
    P("<b>(d) Compare the penalties in Problems C and D</b>", "Part"),
    P("Both terms add a cost to the objective to discourage undesirable behaviour. In Problem C, λΣaᵢ² discourages large model coefficients to help prevent overfitting. In Problem D, βt² discourages excessive drift from the reference language model. In both cases, the penalty balances the main objective against a constraint or preference for a more controlled solution.", "Body2"),
]

story += [
    P("Problem E – Farming Advice (5 points)", "Problem"),
    P("<b>Three opportunities</b>", "Part"),
    P("<b>1. Greater access to advice:</b> Farmers can ask questions in natural language and receive explanations without needing advanced technical knowledge. If local languages are supported, the system may make useful information more accessible to rural communities.", "Body2"),
    P("<b>2. Faster decisions:</b> The system can quickly provide guidance about irrigation, crop treatment and when to seek expert help. This may help farmers respond to common problems sooner.", "Body2"),
    P("<b>3. Access to broad agricultural knowledge:</b> An LLM can explain general farming practices, help farmers understand possible causes of crop problems and suggest questions or information to discuss with an agricultural expert.", "Body2"),
    P("<b>Three risks</b>", "Part"),
    P("<b>1. Incorrect or unsafe advice:</b> LLMs can produce confident but false answers. Incorrect pesticide, treatment or irrigation recommendations could damage crops, harm people or cause financial losses. High-risk advice should be checked by qualified local experts.", "Body2"),
    P("<b>2. Language and local-context limitations:</b> The model may misunderstand local dialects, farming practices, soil, weather or crop varieties. Advice that is reasonable in another region may not be suitable for a particular rural community in Nepal.", "Body2"),
    P("<b>3. Connectivity and privacy:</b> Some farmers may have unreliable internet access or limited access to suitable devices. The system may also collect sensitive farm or personal information, creating privacy and data-security risks if that information is not protected.", "Body2"),
    P("<b>Conclusion</b>", "Part"),
    P("An LLM-based farming assistant could make agricultural information easier to access and support faster decisions. However, it should be tested with local farmers, adapted to local languages and conditions, and designed to refer uncertain or high-risk cases to qualified agricultural experts rather than replacing them.", "Body2"),
]

def add_page_number(canvas, doc):
    canvas.saveState()
    canvas.setFont("Helvetica", 8)
    canvas.setFillColor(HexColor("#6B7C8F"))
    canvas.drawString(18*mm, 12*mm, "IMLC 2026 · Qualification Round")
    canvas.drawRightString(A4[0]-18*mm, 12*mm, f"Page {doc.page}")
    canvas.restoreState()

doc = SimpleDocTemplate(
    out, pagesize=A4,
    rightMargin=18*mm, leftMargin=18*mm,
    topMargin=16*mm, bottomMargin=19*mm,
    title="IMLC 2026 Qualification Round Solutions",
    author="",
)
doc.build(story, onFirstPage=add_page_number, onLaterPages=add_page_number)

print(f"Created: {out}")
print(f"File size: {os.path.getsize(out)/1024:.1f} KB")

from google.colab import files
files.download(out)
```
